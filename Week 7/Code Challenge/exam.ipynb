```python
def temperature_range(df, field):
     Max_temperature_C = df['Max_temperature_C'][df['Field_ID'] == field]
     Min_temperature_C = df['Min_temperature_C'][df['Field_ID'] == field]
     return Max_temperature_C - Min_temperature_C

print(temperature_range(MD_agric_exam, 1458))
print(temperature_range(MD_agric_exam, 1895))
print(temperature_range(MD_agric_exam, 5443))
```

```python
# Get unique crop names
unique_crops = MD_agric_exam['Crop_type'].unique()

# Define recursive function
def sum_crop_lengths(crops, i=0):
    # Base case: if we've processed all crops
    if i == len(crops):
        return 0
    # Recursive step: length of current + sum of remaining
    return len(crops[i]) + sum_crop_lengths(crops, i + 1)

# Compute total
total_sum = sum_crop_lengths(unique_crops)
print(total_sum)
```
